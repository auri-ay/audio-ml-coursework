```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from audioml.shared.mfcc import calc_mel, calc_mfcc
import librosa
import matplotlib.pyplot as plt

x, sr = librosa.load("data/raw/HBStrat/HBStrat_Fm_open_322_ID1_3.wav", sr=44100)

import librosa.display
import numpy as np

fig, ax = plt.subplots()

data = calc_mel(x)
data = librosa.power_to_db(data, ref=np.max)

img = librosa.display.specshow(
    data,
    sr=44100,
    hop_length=441,
    fmin=0,
    fmax=22050,
    y_axis="mel",
    x_axis="time",
    ax=ax
)

ax.set_title("Mel cepstral coefficients spectogram")
ax.set_ylim(0, 22050)

fig.colorbar(img, ax=ax, format="%+2.0f dB")
plt.show()
```
